# **Breast Cancer Prediction with TensorFlow / Keras (Binary Classification)**

**Dataset:** Breast Cancer Wisconsin — 30 numeric features,
predict whether a tumor is *malignant* or *benign*.

**Kaggle Link**: [Breast Cancer Data](https://www.kaggle.com/datasets/uciml/breast-cancer-wisconsin-data)

**The pipeline we'll establish here is the template for every project in this section:**
load → explore → split → scale → build → train → evaluate.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    confusion_matrix, classification_report,
    roc_auc_score, roc_curve
)
print ("tensorFlow", tf.__version__)
```

```python
#set seeds for reproductible results
np.random.seed(42)
tf.random.set_seed(42)

pd.set_option("display.max_columns", None)                  #show all comumns, don't truncate
pd.set_option("display.float_format", lambda x: f'{x:.4f}') # no scientific notation, 4 decimals
np.set_printoptions(suppress=True)                          # no scientific notation for numpy arrays
```

# **Load the data**

```python
df = pd.read_csv("breast_cancer_data.csv")
```

# **Quick exploration**

```python
#drop id and "unnamed: 32" columns
df = df.drop(columns=["id", "Unnamed: 32"])
```

```python
#convert labels to 1 and 0
df["diagnosis"] = df["diagnosis"].map({"B": 0, "M": 1}).astype(int)
```

```python
class_names = ["benign", "malignant"]
```

```python
df.shape
```

```python
df['diagnosis'].value_counts(normalize=True)*100
```

## **Train/Val/test Split**

```python
#split the df into features (X) and target (y)
X = df.drop(columns=["diagnosis"])
y = df["diagnosis"]
```

```python
X.head()
```

```python
y.head()
```

```python
# first peel off the test set
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
#then split the rest (X_train_full) into train (60%) and validation (20%)
X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full, test_size=0.25, stratify=y_train_full, random_state=42
)
print("Train:", X_train.shape)
print("Val:  ", X_val.shape)
print("Test: ", X_test.shape)
```

**Scale the features**

\begin{equation}
z = \frac{x - \mu}{\sigma}
\end{equation}

\mu = mean value
\sigma = the standart deviation

Note: Fit scaler on train only - scaling before splitting leaks test-set mean/std into training (data leakage)

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)
```

Note: transform(not fit_transform) on val/test - they must be scaled with the train set's mean/std; refitting would give each set its own scaling, breaking comparability and leaking their statistics.

**Build the model**

A small fully-connected network:


*   **ReLU** in the hidden layers (the workhorse activation from Section 5)
*   **sigmoid** on the output - it squashes to a value in (0,1) we read as "probabilit of malignant" (our positive class), which is axactly what binary classification needs.



téléchargement (3).png

```python
model = keras.Sequential([
    keras.Input(shape=(X_train.shape[1],)),      # 30 features in
    keras.layers.Dense(16, activation="relu"),
    keras.layers.Dense(8, activation= "relu"),
    keras.layers.Dense(1, activation="sigmoid")   # 1 probability out
])
```

```python
model.summary()
```

```python
#compile the model
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

# **Training**

**Early stopping** watches the validation loss and halts when it stops improving,
then restores the best weights. This is our first line of defense against overfitting.

```python
early_stop = keras.callbacks.EarlyStopping(
    monitor="val_loss", patience=10, restore_best_weights=True
)
```

```python
class PrintEvery10(keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs):
        if (epoch + 1) % 10 == 0:
            print(f"epoch {epoch + 1:3d}  "
                  f"loss {logs['loss']:.4f}  acc {logs['accuracy']:.4f}  "
                  f"val_loss {logs['val_loss']:.4f}  val_acc {logs['val_accuracy']:.4f}")
```

```python
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=32,
    callbacks=[early_stop, PrintEvery10()],
    verbose=0
)
```

```python
print("Trained for", len(history.history["loss"]))
```

**Learning Curves**

```python
hist = history.history

plt.figure(figsize=(11, 4))

plt.subplot(1, 2, 1)
plt.plot(hist["loss"], label="train")
plt.plot(hist["val_loss"], label="val")
plt.title("Loss"); plt.xlabel("epoch"); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(hist["accuracy"], label="train")
plt.plot(hist["val_accuracy"], label="val")
plt.title("Accuracy"); plt.xlabel("epoch"); plt.legend()

plt.tight_layout()
plt.show()
```

# **Evaluate on the test set**

\>Note: This is the first and only time we use the test set.

```python
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print("Test loss:", round(test_loss, 3))
print("Test accuracy:", round(test_acc, 3))
```

```python
y_prob = model.predict(X_test, verbose=0).ravel()
y_pred = (y_prob >= 0.5).astype(int)

print("First 5 probabilities:", y_prob[:5].round(3))
print("First 5 predictions:  ", y_pred[:5])
```

**Confusion Matrix**

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(4.5, 4))
plt.imshow(cm, cmap="Blues")
plt.title("Confusion Matrix")
plt.xlabel("Predicted"); plt.ylabel("Actual")
plt.xticks([0, 1], ["benign", "malignant"])
plt.yticks([0, 1], ["benign", "malignant"])
for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i, j], ha="center", va="center", fontsize=13)
plt.colorbar()
plt.tight_layout()
plt.show()
```

**Precision, recall, F1**

```python
print(classification_report(y_test, y_pred, target_names = class_names))
```

# **ROC curve and AUC**

```python
auc = roc_auc_score(y_test, y_prob)
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

plt.figure(figsize=(5, 4))
plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
plt.plot([0, 1], [0, 1], "--", color="gray")
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curve"); plt.legend()
plt.tight_layout()
plt.show()
```

# **Save the model and scaler**

```python
import joblib

# save the model
model.save("breast_cancer_model.keras")

# saving the scaler
joblib.dump(scaler, "scaler.joblib")

print("Saved: breast_cancer_model.keras, scaler.joblib")
```

## Wrap-up

What we built, and where the Section 5 theory showed up:

- **sigmoid + binary cross-entropy** turned a network into a binary classifier (steps 6–7)
- **scaling** got every feature onto a comparable range so training was stable (step 5)
- **early stopping + the learning curves** kept overfitting in check (steps 7–8)
- **confusion matrix, precision/recall, ROC-AUC** told us *how* the model is right or
  wrong — far more than a single accuracy number (steps 10–12)
